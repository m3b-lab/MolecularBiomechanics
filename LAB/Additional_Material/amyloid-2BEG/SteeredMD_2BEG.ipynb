# Steered Molecular Dynamics (SMD)

## Theory Recap

**Steered Molecular Dynamcis (SMD)** applies external steering forces, applying a **constraint** (e.g. a harmonic potential), that moves along a prescribed path in the configuration space in the right direction to accelerate processes that otherwise, due to energy barriers, are too slow, overtaking the limitation of the nanosecond time scale usual in classical molecular dynamics, which is seldom long enough to observe relevant processes.

In order to guarantee that the reaction coordinate closely follows the **constraint positions**, the force constant F must be sufficiently large, which is called the **stiff-spring approximation theory**:


$$
\begin{aligned}
& F(t)=2 k(v t-s(t)) \\
& W(x(t))=\int_0^{x(t)} F(t) d x(t)
\end{aligned}
$$

<center><img src="imgs/SMD_scheme.png" width="500"></center>

In a typical investigation of a molecular process, the **reaction path**, along which the process proceeds in the configurational space, is identified (or hypothesized) and the process progress is described by the **reaction coordinate**.

The **potential of mean force (PMF)** is basically the ***free energy profile along the reaction coordinate***. PMF is determined through the Boltzmann-weighted average over all degrees of freedom other than the reaction coordinate. ​

<center><img src="imgs/PMF.png" width="500"></center>

SMD simulations are non-equilibrium processes, but PMF is an equilibrium property. The **Jarzynski’s equality** ***connects equilibrium and non-equilibrium***

$$\left<e^{-\beta {W}}\right>=e^{-\beta \Delta G} \Longleftrightarrow \Delta G=-\frac{1}{\beta} \ln \left<e^{-\beta {W}}\right>$$

### Step 8 - Steered Molecular Dynamics (SMD)

The idea is to pull one chain of the fibril away applying an external force while keeping fixed the position of the adjacent chain using position restraints. 

<center><img src="imgs/SystemScheme.png" width="1200"></center>

We want to pull one of the chain of the amyloid fibril along a chosen **reaction coordinate**. 

This is equivalent to ideally attach a **spring** with a certain **elastic constant** at the end of the fibril and then apply a displacemnt with a specific **force** and **velocity**.

For this example, the reaction coordinate is the **z-axis**. To generate these configurations, we must pull peptide A away from the protofibril. We will pull over the course of 500 ps of MD, saving snapshots every 1 ps.

How can we compute the work done by the pulling force?

In constant-velocity pulling, the anchor of the spring moves at a fixed speed $v$ (`pull_coord1_rate` in `data/pull.mdp`). The work done on the system by this pulling protocol is

$$W(t) = v\int_0^t F(t')\,dt'$$

where $F$ is the spring force that GROMACS writes to `pull_pullf.xvg`. This is the work that enters Jarzynski's equality below.

We plot it against the position of the anchor, $\lambda(t) = \lambda_0 + v\,t$, not against the measured distance: the pulled chain lags behind the anchor by $F/k$ whenever the spring is stretched.

We are also interested in evaluating the energy difference $\Delta G$ between state A (bound state) and state B (un-bound state). 

Remember the **Jarzinski equality**: 

$$\left<e^{-\beta {W}}\right>=e^{-\beta \Delta G} \Longleftrightarrow \Delta G=-\frac{1}{\beta} \ln \left<e^{-\beta {W}}\right>$$


We can extract **equilibrium information ΔG** from the ensemble of **non-equilibrium (finite-time) measurements**.

**Note:** the average $\left<\cdot\right>$ runs over many independent pulls, each started from an equilibrated configuration. From the single pull above we can compute the work, but not $\Delta G$.

```python
# evaluate the work done by the pulling protocol
import numpy as np

rate = 0.01 # nm/ps, pull_coord1_rate in data/pull.mdp

# W(t) = rate * integral of the spring force over time (trapezoidal rule)
work = rate * np.concatenate([[0], np.cumsum(0.5*(force[1:] + force[:-1])*np.diff(time))])

# position of the spring anchor: it starts at the initial distance and moves at constant speed
anchor = displacement[0] + rate*time

plt.plot(anchor,work)
plt.xlabel("Spring anchor position (nm)",size=14)
plt.ylabel("Work (kJ/mol)",size=14)
plt.title("Work done by the pulling protocol")
plt.tight_layout()
# uncomment the following line to save the figure
#plt.savefig("work-anchor.png",format="png", dpi=300)
```

```python
# print the results
print(f"Total work done by the force {work[-1]:.2f} kJ/mol")
```

**Why a single pull cannot give $\Delta G$.** With one trajectory the average $\left<e^{-\beta W}\right>$ has a single term, so $-\frac{1}{\beta}\ln e^{-\beta W} = W$ exactly, whatever the value of $\beta$: the "free energy" would just be the work under another name.

The work of a finite-speed pull also contains a dissipated part that is lost as heat, and one value of $W$ cannot tell you how large it is. Only the distribution of $W$ over many independent pulls can. The next step does exactly that.

```python
import matplotlib.pyplot as plt 
import numpy as np
import glob

#load data
data_f = []
data_d = []
forces = []
displacements = []

# create figure for plot
fig,ax = plt.subplots(1,1)

# load data for each simulation found in the folder
for pull_dir in sorted(glob.glob("03-multiple_pull/pull_*")):

    # load data
    data_f = np.loadtxt(pull_dir + "/pull_pullf.xvg",comments=["#","@"])
    data_d = np.loadtxt(pull_dir + "/pull_pullx.xvg",comments=["#","@"])

    # retrieve data
    time=data_f[:,0] 
    force = data_f[:,1]
    displacement = data_d[:,1]

    # append data to lists
    forces.append(force)
    displacements.append(displacement)

    # plot with matplotlib
    plt.plot(displacement,force)

# convert to numpy arrays
forces = np.array(forces)
displacements = np.array(displacements)

plt.xlabel("Displacement (nm)",size=14) 
plt.ylabel("Force ($kJmol^{-1}nm^{-1}$)",size=14) 
plt.title("Force-Displacement")
plt.tight_layout()
```

```python
temperature = 310 # K, must match ref_t in data/pull.mdp
boltzmann = 0.008314 # kJ/mol/K, same energy unit as the work
beta = 1/(boltzmann*temperature)
rate = 0.01 # nm/ps, pull_coord1_rate in data/pull.mdp

# all pulls share the same output times, so the anchor position is the same at each time
anchor = np.mean(displacements[:,0]) + rate*time

works = []

# cycle of simulations
for force in forces:

    # protocol work: W(t) = rate * integral of the spring force over time
    work = rate * np.concatenate([[0], np.cumsum(0.5*(force[1:] + force[:-1])*np.diff(time))])

    works.append(work)

    plt.plot(anchor, work, color="gray",ls='dotted')

works = np.array(works)

# calculate the free energy with Jarzynski: delta_g = -1/beta * ln < exp(-beta*W) >, averaged at fixed anchor position
# beta*W is in the hundreds or thousands, so exp(-beta*W) underflows to 0 and the log returns inf.
# Factoring out the smallest work keeps the exponentials in range without changing the result.
w_min = np.min(works, axis=0)
delta_g = w_min - 1/beta*np.log(np.mean(np.exp(-beta*(works - w_min)), axis=0))

plt.plot(anchor, delta_g, color="red",ls='solid', label="PMF")
plt.xlabel("Spring anchor position ($nm$)")
plt.ylabel("Energy $kJ*mol^{-1}$")
plt.legend()
plt.tight_layout()
```

Look at the first 2 nm: the chain has not detached yet, so almost all the work goes into stretching the spring and grows roughly like $\frac{1}{2}k(\lambda-\lambda_0)^2$. With $k = 1000$ kJ mol$^{-1}$ nm$^{-2}$ the spring stretches by up to 2 nm, far too soft for this curve to be read as the free-energy profile of the chain itself. That reading needs a spring stiff enough to keep the chain close to the anchor (the stiff-spring condition of Park & Schulten, J. Chem. Phys. 2004).

### Mean work vs Jarzynski: the dissipation gap

Compare the Jarzynski PMF with the plain average of the work $\left<W\right>$. Since $e^{-x}$ is convex, Jensen's inequality gives $\left<e^{-\beta W}\right> \geq e^{-\beta \left<W\right>}$, hence

$$\Delta G \leq \left<W\right>$$

The gap $\left<W\right> - \Delta G$ is the work dissipated as heat. It grows with the pulling speed.

The exponential average is dominated by the rare **low-work** pulls, not by the typical ones: that is the whole content of Jarzynski's equality, and the reason it needs many replicas. When the spread of the work $\sigma_W$ is much larger than $k_BT$, a handful of pulls cannot sample the low-work tail. The estimate then just follows the lowest-work replica and overestimates $\Delta G$.

**Try it:** set `boltzmann = 8.314` (J/mol/K instead of kJ/mol/K) in the cell above and run both cells again. Which curve does the PMF collapse onto, and why?

```python
mean_work = np.mean(works, axis=0)

plt.plot(anchor, mean_work, color="blue", label=r"$\left<W\right>$")
plt.plot(anchor, delta_g, color="red", label=r"$\Delta G$ (Jarzynski)")
plt.fill_between(anchor, delta_g, mean_work, color="gray", alpha=0.3, label="dissipated work")
plt.xlabel("Spring anchor position ($nm$)")
plt.ylabel("Energy $kJ*mol^{-1}$")
plt.legend()
plt.tight_layout()

kT = boltzmann*temperature
sigma_w = np.std(works[:,-1], ddof=1)
print(f"Replicas: {len(works)}")
print(f"<W> = {mean_work[-1]:.1f} kJ/mol, ΔG = {delta_g[-1]:.1f} kJ/mol, dissipated = {mean_work[-1]-delta_g[-1]:.1f} kJ/mol")
print(f"Spread of the work: σ_W = {sigma_w:.1f} kJ/mol = {sigma_w/kT:.0f} kT")
```

### The work distribution

The physics is in the **distribution** of the work, not in any single value. Where do $\left<W\right>$ and $\Delta G$ fall relative to the histogram of the final work? How many replicas contribute to the Jarzynski estimate in practice?

```python
final_work = works[:,-1]

plt.hist(final_work, bins="auto", color="gray", edgecolor="black")
plt.axvline(mean_work[-1], color="blue", label=r"$\left<W\right>$")
plt.axvline(delta_g[-1], color="red", label=r"$\Delta G$ (Jarzynski)")
plt.xlabel("Work ($kJ*mol^{-1}$)")
plt.ylabel("Number of pulls")
plt.legend()
plt.tight_layout()
```
